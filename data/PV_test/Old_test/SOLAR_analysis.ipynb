```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv("SOLAR_ina226_log_20260423_121933.csv")

# Use raw data
V = df["bus_V"]

# If your current is negative because of sensor direction, use abs()
I = df["current_mA"].abs()

# Use measured power from the CSV
P = df["power_mW"]

# If you would rather calculate power yourself, use this instead:
# P = V * I

print(df[["bus_V", "current_mA", "power_mW"]].head())

# Voltage vs Current
plt.figure(figsize=(7,5))
plt.plot(I, V, ".")
plt.xlabel("Current [mA]")
plt.ylabel("Voltage [V]")
plt.title("Voltage vs Current")
plt.grid(True)
plt.tight_layout()
plt.show()

# Power vs Current
plt.figure(figsize=(7,5))
plt.plot(I, P, ".")
plt.xlabel("Current [mA]")
plt.ylabel("Power [mW]")
plt.title("Power vs Current")
plt.grid(True)
plt.tight_layout()
plt.show()
```
