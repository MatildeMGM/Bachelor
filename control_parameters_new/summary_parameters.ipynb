## 2. Critical notes

- Demand thresholds are in mW, not W.
- Price thresholds are in DKK/kWh after converting from DKK/MWh.
- The battery is scaled by energy to 100 mWh for the EMS demo.
- The PEM power limit comes from the polarization curve while staying above the voltage cutoff.
- The PEM hydrogen end reading belongs to the full-test endpoint, not the EMS cutoff voltage.
- PEM charging is treated as an energy accumulation problem, not as a fixed PEM charging current.
- One useful PEM output event is defined as 60 s at the selected usable PEM discharge power, requiring about 12.8 J of PEM input charge energy.
- The PV measurements can supply the 100 mW load and can accumulate the derived useful PEM charge energy in the tested ramps.
